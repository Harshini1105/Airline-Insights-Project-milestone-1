```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("flights.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
print(df.columns)
```

```python
#univariate analysis/ chart 1/top airlines
plt.figure(figsize=(8,5))
df['carrier'].value_counts().head(10).plot(kind='bar')
plt.title("Top Airlines by Number of Flights")
plt.xlabel("Carrier")
plt.ylabel("Flight Count")
plt.show()
```

```python
#flight by month
plt.figure(figsize=(8,5))
df['month'].value_counts().sort_index().plot(kind='bar')
plt.title("Flights per Month")
plt.xlabel("Month")
plt.ylabel("Number of Flights")
plt.show()
```

```python
#top origin airports
plt.figure(figsize=(8,5))
df['origin'].value_counts().plot(kind='bar')
plt.title("Flights by Origin Airport")
plt.show()
```

```python
#arrival  delay distribution
plt.figure(figsize=(8,5))
df['arr_delay'].dropna().plot(kind='hist', bins=50)
plt.title("Arrival Delay Distribution")
plt.xlabel("Delay (Minutes)")
plt.show()
```

```python
#airline vs average delay
plt.figure(figsize=(10,5))
df.groupby('carrier')['arr_delay'].mean().sort_values().plot(kind='bar')
plt.title("Average Arrival Delay by Airline")
plt.ylabel("Average Delay (Minutes)")
plt.show()
```

**Observation:**  
Some airlines show consistently higher average delays compared to others.

```python
#chart2/ month vs average delay
plt.figure(figsize=(8,5))
df.groupby('month')['arr_delay'].mean().plot(kind='line', marker='o')
plt.title("Average Arrival Delay by Month")
plt.ylabel("Average Delay (Minutes)")
plt.show()
```

```python
#top airlines by flight
plt.figure(figsize=(8,5))
df['carrier'].value_counts().head(10).plot(kind='bar')
plt.title("Top Airlines by Number of Flights")
plt.show()
```

```python
#depature vs arrival delay relationship
plt.figure(figsize=(6,5))
plt.scatter(df['dep_delay'], df['arr_delay'], alpha=0.3)
plt.title("Departure Delay vs Arrival Delay")
plt.xlabel("Departure Delay")
plt.ylabel("Arrival Delay")
plt.show()
```

```python
#route analysis
# Top 10 busiest routes
df['route'] = df['origin'] + "-" + df['dest']

plt.figure(figsize=(8,5))
df['route'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Busiest Routes")
plt.show()
```

```python
#correlation heatmap
import seaborn as sns

plt.figure(figsize=(8,6))

sns.heatmap(
    df[['arr_delay','dep_delay','month','hour']].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Between Flight Variables")
plt.show()
```

```python
#delay by time of day
plt.figure(figsize=(8,5))
df.groupby('hour')['arr_delay'].mean().plot(kind='line', marker='o')
plt.title("Average Arrival Delay by Hour")
plt.ylabel("Average Delay (Minutes)")
plt.xlabel("Hour of Day")
plt.show()
```
