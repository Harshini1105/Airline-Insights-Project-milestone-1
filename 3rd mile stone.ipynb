```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\prakash\OneDrive\Desktop\Air fly\cleaned_dataset.csv")
```

```python
df['Route'] = df['Origin'] + "-" + df['Dest']
```

```python
top_routes = df['Route'].value_counts().head(10)

plt.figure()
top_routes.plot(kind='bar')
plt.title("Top 10 Flight Routes")
plt.xlabel("Route")
plt.ylabel("Number of Flights")
plt.show()
```

```python
pivot_table = df.pivot_table(
    values='ArrDelay',
    index='Origin',
    columns='Dest',
    aggfunc='mean'
)

plt.figure(figsize=(10,6))
sns.heatmap(pivot_table, cmap="coolwarm")
plt.title("Average Delay Heatmap")
plt.show()
```

```python
busiest_airports = df['Origin'].value_counts().head(10)

plt.figure()
busiest_airports.plot(kind='bar')
plt.title("Busiest Airports")
plt.xlabel("Airport")
plt.ylabel("Flights")
plt.show()
```

```python
cancel_month = df.groupby('Month')['Cancelled'].sum()

plt.figure()
cancel_month.plot(kind='line')
plt.title("Monthly Flight Cancellations")
plt.xlabel("Month")
plt.ylabel("Cancelled Flights")
plt.show()
```

```python
cancel_reasons = df[['CarrierDelay','WeatherDelay','NASDelay','SecurityDelay']].sum()

plt.figure()
cancel_reasons.plot(kind='bar')
plt.title("Delay Causes")
plt.show()
```

```python
winter = df[df['Month'].isin([12,1,2])]

winter_delay = winter['ArrDelay'].mean()

print("Average winter delay:", winter_delay)
```
